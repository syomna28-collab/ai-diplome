```python
numbers_of_members=input("enter the number of teams members")
number_of_months=input("enter number of months")
monthly_salary=input("enter the monthly salary")
cloud_cost=input("enter total cloud cost for the whole project")
numbers_of_members=int(numbers_of_members)
number_of_months=float(number_of_months) #3mltaha float mesh int ashan momken yb2a shahren w nos masalan
monthly_salary=float(monthly_salary)
cloud_cost=float(cloud_cost)
salary_cost=numbers_of_members*number_of_months*monthly_salary
subtotal=salary_cost+cloud_cost
VAT=subtotal*0.14
total=subtotal+VAT
print(f"salary cost:{salary_cost:,.2f}")
print(f"subtotal cost:{subtotal:,.2f}")
print(f"VAT:{VAT:,.2f}")
print(f"total:{total:,.2f}")
```
