```python
customers=["Ahmed", "Sara", "Omar", "Mona",
"Ali", "Nour", "Youssef", "Hana",
"Khaled", "Mariam", "Hassan", "Laila"]
scores=[91, 45, 78, 66, 88, 30,
72, 95, 51, 60, 84, 39]
risk_levels=[]
best=scores[0]
worst=scores[0]
best_customer=customers[0]
worst_customer=customers[0]
h=0
l=0
m=0
total=0
for customer,score in zip(customers,scores) :
    if 85<=score <=100 :
        risk="high"
        h+=1
    elif 60<=score<=84 :
        risk="medium"
        m+=1
    else :
        risk="low"
        l+=1
    risk_levels.append(risk)
    total+=score
    if score<best :
     best=score
     best_customer=customer
    if score>worst:
     worst=score
     worst_customer=customer
average=total/len(scores)
while True :
 print("="*10,"menu","="*10)
 print("1-Show All Customers")
 print("2-Search for a Customer")
 print("0-Exit")

 choice=input("enter your choice :")
 if choice=="1":       
    print("="*10,"customer risk report","="*10)
    print()
    print("name"," ","score"," ", "risk")
    print("-"*20)
    for customer,score,risk in zip(customers,scores,risk_levels):
      print(customer," ",score," ",risk)
    print("-"*20)
    print("="*10,"summary","="*10)
    print("high risk:",h)
    print("medium risk:",m)
    print("low risk:",l)
    print("average score:",f"{average:.2f}")
    print("best customer:",best_customer)
    print("worst customer:",worst_customer)
 elif choice=="2":
       customer_name=input("enter customer name:")
       found=False
       for customer,score,risk in zip(customers,scores,risk_levels):
            if customer==customer_name:
                print("Name:",customer)
                print("Score:",score)
                print("Risk:",risk)
                found=True
                break
            if found==False:
             print("Customer not found")
 elif choice=="0":
    break
 else:
    print("invalid choise")
```
